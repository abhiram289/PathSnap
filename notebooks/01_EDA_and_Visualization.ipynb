## 2. Inspecting Sample Trip 00000001 (Track, Nodes, Arcs, Route)

```python
trip_id = "00000001"
trip_dir = os.path.join("Student folder", "train", trip_id)
track = pd.read_csv(os.path.join(trip_dir, f"{trip_id}.track"), sep=r'\s+', header=None, names=['lon', 'lat', 'time'])
nodes = pd.read_csv(os.path.join(trip_dir, f"{trip_id}.nodes"), sep=r'\s+', header=None, names=['lon', 'lat'])
arcs = pd.read_csv(os.path.join(trip_dir, f"{trip_id}.arcs"), sep=r'\s+', header=None, names=['u', 'v'])
route = pd.read_csv(os.path.join(trip_dir, f"{trip_id}.route"), sep=r'\s+', header=None, names=['arc_id'])

print(f"Track observations: {len(track)}")
print(f"Road network nodes: {len(nodes):,}")
print(f"Road network arcs:  {len(arcs):,}")
print(f"Ground-truth route arcs: {len(route)}")
track.head()
```

## 3. Sampling Interval Distribution

```python
dt = track['time'].diff().dropna()
print(f"Median delta: {dt.median():.2f}s | Max delta: {dt.max():.2f}s | % exactly 1s: {(dt == 1.0).mean()*100:.2f}%")

plt.figure(figsize=(8, 4))
dt_counts = dt.value_counts().sort_index()
dt_counts[dt_counts.index <= 5].plot(kind='bar', color='#3498db', edgecolor='black')
plt.title("GPS Sampling Interval Distribution (Trip 00000001)", fontweight='bold')
plt.xlabel("Delta Time (seconds)")
plt.ylabel("Count")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
